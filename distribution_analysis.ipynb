The dataset is illustrative sample data.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'student_study_hours.csv')
values = df['Study_Hours']
df.head()
```

```
plt.figure(figsize=(9, 5))
plt.hist(values, bins=8, edgecolor='black')
plt.title('Distribution of Student Study Hours')
plt.xlabel('Study Hours')
plt.ylabel('Number of Students')
plt.grid(axis='y', alpha=0.25)
plt.tight_layout()
plt.show()
```

```
print('Count:', len(values))
print('Mean:', round(values.mean(), 2))
print('Median:', round(values.median(), 2))
print('Minimum:', values.min())
print('Maximum:', values.max())
print('Standard deviation:', round(values.std(), 2))
```

## Observations
1. Most values are in the lower-to-middle range.
2. Values range from 1 to 20 hours, showing a wide spread.
3. A few high values extend the right tail, suggesting positive skew.

These conclusions refer only to illustrative sample data.
